```python
# membuat data untuk diolah
x = [1,2,3,4,5]
y = [2,4,6,8,10]

# membuat plot garis
plt.plot(x,y, 'g--', linewidth=2.0)

# menambahkan label pada sumbu x dan y
plt.xlabel('Sumbu X')
plt.ylabel('Sumbu Y')

# menambahkan judul pada plot
plt.title('Contoh Plot Garis')

# menetukan ukuran plot
plt.figure(figsize=(5,5))

# menampilkan plot
plt.show()
```

#### Menambahkan Anotasi

```python
# membuat data untuk diolah
x = [1,2,3,4,5]
y = [2,4,6,8,10]

# membuat plot garis
plt.plot(x,y, '-o')

# menambahkan label pada sumbu x dan y
plt.xlabel('Sumbu X')
plt.ylabel('Sumbu Y')

# menambahkan judul pada plot
plt.title('Contoh Plot Garis dengan Anotasi')

# menambahkan anotasi
plt.annotate('Puncak', xy=(5,10), xytext=(3,20), arrowprops=dict(facecolor='black', shrink=0.05))

# menampilkan plot
plt.show()
```

#### Bar Chart

```python
# membuat data untuk diolah
nama_kota = ['Jakarta', 'Bandung', 'Surabaya', 'Medan', 'Bekasi']
jumlah_penduduk = [10.4, 7.8, 3.5, 2.3, 5.2]

# membuat bar chart
plt.bar(nama_kota, jumlah_penduduk)
plt.xlabel('Nama Kota')
plt.ylabel('Jumlah Penduduk (juta jiwa)')
plt.title('Jumlah Penduduk Berdasarkan Kota')

# menampilkan bar chart
plt.show()
```

#### Pie Chart

```python
# Membuat data untuk diolah
makanan = ['Nasi Goreng', 'Sate', 'Bakso', 'Gado-gado', 'Bubur Ayam']
persentase = [25, 20, 15, 30, 10]

# Membuat pie chart
plt.pie(persentase, labels=makanan, startangle=90,
        counterclock=False, autopct='%1.1f%%', shadow=True)
plt.axis('equal')
plt.title('Persentase Pembelian Makanan')

# Menampilkan pie chart
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data yang akan ditampilkan
labels = ['A', 'B', 'C', 'D']
sizes = [10, 20, 30, 40]
colors = ['lightcoral', 'lightblue', 'lightgreen', 'yellowgreen']

# Membuat figure dan axis
fig, ax = plt.subplots()

# Membuat pie chart
ax.pie(sizes, labels=labels, colors=colors,
       startangle=90, autopct='%1.1f%%', shadow=True)

# Tambahkan legenda pada pie chart
ax.legend(loc='best')

# Membuat tampilan chart agar lebih rapi
ax.axis('equal')

# Tampilkan plot
plt.show()
```

#### Scatterplot

```python
import matplotlib.pyplot as plt
import numpy as np

# Membuat data untuk diolah
np.random.seed(0)
cluster1 = np.random.normal(50, 5, (100, 2))
cluster2 = np.random.normal(30, 5, (100, 2))
cluster3 = np.random.normal(70, 5, (100, 2))
cluster4 = np.random.normal(10, 5, (100, 2))

# Menggabungkan data
data = np.concatenate((cluster1, cluster2, cluster3, cluster4))

# Membuat scatterplot
colors = ['red', 'blue', 'green', 'purple']
for i in range(4):
    cluster = data[i*100:(i+1)*100]
    x = cluster[:, 0]
    y = cluster[:, 1]
    plt.scatter(x, y, color=colors[i], s=30)

plt.title('Scatterplot 4 Cluster dengan Variasi Warna')
plt.xlabel('X')
plt.ylabel('Y')

# Menampilkan scatterplot
plt.show()
```

#### Plot Multigrafik

```python
import matplotlib.pyplot as plt
import numpy as np

# Membuat data untuk diolah
x = np.arange(0, 10, 0.2)
y = np.sin(x)
z = np.cos(x)

# Membuat subplot pertama
plt.subplot(2, 1, 1)
plt.plot(x, y, 'r')
plt.title('Grafik Sin(x)')
plt.xlabel('Nilai x')
plt.ylabel('Nilai y')

# Membuat subplot kedua
plt.subplot(2, 1, 2)
plt.plot(x, z, 'b')
plt.title('Grafik Cos(x)')
plt.xlabel('Nilai x')
plt.ylabel('Nilai y')

# Menampilkan multi plot
plt.subplots_adjust(bottom=0.1, right=0.8, top=2)
plt.show()
```

#### Plot 3D

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Membuat data
x = np.linspace(-2, 2, 200)
y = np.linspace(-2, 2, 200)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) + np.cos(Y)

# Membuat figure dan axis 3D
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plot garis 3D
ax.plot_surface(X, Y, Z, cmap='viridis')

# Tampilkan plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Membuat data
N = 1000
x = np.random.rand(N)
y = np.random.rand(N)
z = np.random.rand(N)

# Membuat figure dan axis 3D
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plot scatter 3D
ax.scatter(x, y, z, c=z, cmap='viridis')

# Tampilkan plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Membuat data untuk diolah
X = np.linspace(-np.pi, np.pi, 256, endpoint=True)
C = np.cos(X)
S = np.sin(X)

# Membuat plot 3D
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.plot(X, C, S)

# Menambahkan label pada sumbu x, y, dan z
ax.set_xlabel('Sumbu X')
ax.set_ylabel('Sumbu Y')
ax.set_zlabel('Sumbu Z')

# Menambahkan judul pada plot
plt.title('Contoh Plot 3D')

# Menentukan ukuran plot
plt.figure(figsize=(5, 5))

# Menampilkan plot
plt.show()
```
